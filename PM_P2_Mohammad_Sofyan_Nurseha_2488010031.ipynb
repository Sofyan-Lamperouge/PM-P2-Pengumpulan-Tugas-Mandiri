**C. Langkah Latihan**

Langkah 1 - Mengenali Fitur & Label

```python
import pandas as pd
data = {'luas':[36,70,45,90,54],'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]} #Label

df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']] #Fitur
y = df['harga']      #label
```

Langkah 2 - Klasifikasi vs Regresi

```python
# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']
```

Langkah 3 - Pembagian data latih & uji

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size = 0.4,random_state = 42)
print('Latih:', len(X_train), 'Uji:', len(X_test))
```

**E. Refleksi**

1. Apa perbedaan supervised, unsupervised, dan reinforcement learning?


  Kalau Supervised model dilatih dengan menggunakan data yang mempunyai label atau jawaban.

  Sedangkan Unsupervised model dilatih dengan menggunakan data yang tidak mempunyai label atau jawaban, tujuannya untuk mencari pola tersembunyi pada data.

  Lalu untuk Reinforcement learning model dilatih dengan reward dan punishment, jika jawaban akurat maka point bertambah sedangkan jika jawaban tidak akurat maka point berkurang.

2. Mengapa pembagian data latih dan uji penting?

  Karena dengan adanya pembagian data kita bisa tahu apakah model yang kita buat sudah bisa menentukan jawaban sendiri dengan akurat apa belum. Data latih digunakan sebagai materi untuk belajar sedangkan data uji digunakan sebagai ujian dengan data yang baru atau data yang belum digunakan sebelumnya.


3. Mengapa alur kerja ML disebut iteratif?

  Karena dalam membuat sebuah model ML jarang sekali berhasil pada percobaan pertama. Jika model belum akurat bisa dilakukan pengulangan pada tahap sebelumnya sampai mendapatkan model yang akurat.
